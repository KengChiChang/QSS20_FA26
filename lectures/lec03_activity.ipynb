## Part 1: choose a cleaning rule

A count depends on what the program treats as a word. Compare three plausible rules before choosing one. The edge cases include capitalization, punctuation, apostrophes, a substring, and a number.

```python
from pathlib import Path
import re

# The notebook lives in lectures/; the data live one level up in data/.
ROOT = Path("..") if Path("../data").exists() else Path(".")
FILES = sorted((ROOT / "data/raw/federalist").glob("fp*.txt"))

edge_cases = [
    "Nearby, by the sea, BY design.",
    "The nation's interest isn't one person's interest.",
    "Article 2 follows Article II.",
]

def rule_a(text):
    return text.lower().split()

def rule_b(text):
    return re.sub(r"[^a-z]+", " ", text.lower()).split()

def rule_c(text):
    return re.findall(r"[a-z]+(?:'[a-z]+)?|[0-9]+", text.lower())

for number, text in enumerate(edge_cases, start=1):
    print(f"case {number}: {text}")
    print("  rule A:", rule_a(text))
    print("  rule B:", rule_b(text))
    print("  rule C:", rule_c(text))
```
